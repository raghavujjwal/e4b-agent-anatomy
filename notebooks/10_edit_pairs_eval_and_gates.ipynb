# 10 - Edit pairs: serve the three arms, measure the edit success rate, run the gates (GPU runtime)

Plan: `docs/PLAN_EDIT_PAIRS.md`. Needs the adapters of notebook 09 on Drive (`MyDrive/gemma4-adapters/pair_dpo`, `pair_rft`, `pair_dpo_nll`). One server holds the W4A16 base (what the scorer serves) and the adapters, and this notebook asks:
- **G2** Does vLLM load the adapters (rank 32) next to the base and answer requests? Is each one actually applied?
- **Primary outcome.** On 300 held-out edit steps (2 generations each, shipped sampling): how often is a call applied, exactly, at the right place (`good`)? The difference to the base, with a paired bootstrap over steps.
  Guards: it still reproduces the gold edit as often, it still edits at about the same share of steps, it writes no more unreadable calls.
- **G3, G4, G4b, G5.** The same four gates as notebook 08, on prefixes of the base model's own runs (does it still choose the same kind of step, repeat a failed edit, give up after one, write well-formed calls?).
- **Format probe** (report only): the 55 prefixes before a call the framework refused, 3 samples each.
- **The pre-registered reading** (`edit_pairs.py decide`): one line per arm. **A pass only means "not broken and better at this one step".** It says nothing about resolved tasks; that needs the agent run (notebook 03 with the adapter).

Use a **GPU runtime**. About 15 minutes of setup (vLLM install, model copy), then about 30-40 minutes of generation: roughly 8-12 units. **Note the unit balance before section 3 and after section 9.**

## 5. Start the server with the adapters

Flags follow `HARNESS_README` section 3.1 plus LoRA the way the scorer does it: `enable_lora`, `max_loras` = number of adapters, `max_lora_rank` = the largest adapter rank. **If the server exits or refuses an adapter, this is the answer to G2:**
paste the error (the cell prints the end of the log). One GPU instead of 4 x L4 with tensor parallel 4, and a newer vLLM than the scorer's.

```python
REASONING_CONFIG = json.dumps({"reasoning_start_str": "<|channel>", "reasoning_end_str": "<channel|>"})   # what adk_submission.server.VllmServer adds for --reasoning-parser gemma4
base = f"http://127.0.0.1:{PORT}/v1"


def start_server(with_reasoning_config):
    cmd = [f"{VENV}/bin/vllm", "serve", LOCAL_MODEL, "--served-model-name", SERVED_NAME, "--port", str(PORT),
           "--max-model-len", "32768", "--gpu-memory-utilization", "0.90",
           "--enable-auto-tool-choice", "--tool-call-parser", "gemma4", "--reasoning-parser", "gemma4"]
    if with_reasoning_config:
        cmd += ["--reasoning-config", REASONING_CONFIG]
    cmd += ["--enable-lora", "--max-loras", str(len(ADAPTERS)), "--max-lora-rank", str(max_rank), "--lora-modules", *LORA_ARGS]
    print(" ".join(cmd))
    proc = subprocess.Popen(cmd, stdout=open("/content/vllm.log", "w"), stderr=subprocess.STDOUT)
    t0 = time.time()
    while True:
        if proc.poll() is not None:
            return proc, False, time.time() - t0
        try:
            urllib.request.urlopen(f"{base}/models", timeout=3).read()
            return proc, True, time.time() - t0
        except Exception:
            if time.time() - t0 > 1500:
                proc.kill()
                raise TimeoutError("server not ready after 25 min: " + open("/content/vllm.log").read()[-3000:])
            time.sleep(10)


server, ok, secs = start_server(True)
if not ok and "reasoning" in open("/content/vllm.log").read().lower() and "config" in open("/content/vllm.log").read().lower():
    print("The server stopped; its log mentions the reasoning config. Starting once more without --reasoning-config (the checks send token ids and do not use thinking budgets).")
    server, ok, secs = start_server(False)
if not ok:
    print("SERVER EXITED with code", server.returncode)
    print(open("/content/vllm.log").read()[-5000:])
    raise SystemExit
print("ready after %.0fs" % secs)
print("served models:", [m["id"] for m in json.load(urllib.request.urlopen(f"{base}/models"))["data"]])
print([l for l in open("/content/vllm.log").read().splitlines() if "lora" in l.lower()][-8:])
```

## 6. G2: the adapters load, answer 10 requests, and are applied

Ten chat requests with a tool declared, to each adapter. Then the same one-token request at temperature 0 to the base model and to each adapter: if the first-token log-probabilities are **identical**, the adapter is not being applied
(a loaded-but-ignored adapter would otherwise pass every later check trivially). Any difference at all passes this check; the adapters are rank 32 / alpha 64, so served at LoRA scale 2.0, the scale they were trained at.

```python
def chat(model, messages, tools=None, max_tokens=256, temperature=1.0, **extra):
    body = {"model": model, "messages": messages, "max_tokens": max_tokens, "temperature": temperature, "top_p": 0.95, "top_k": 64,
            "chat_template_kwargs": {"enable_thinking": False}, **extra}
    if tools:
        body["tools"] = tools
    req = urllib.request.Request(f"{base}/chat/completions", data=json.dumps(body).encode(), headers={"Content-Type": "application/json"})
    with urllib.request.urlopen(req, timeout=300) as r:
        return json.load(r)


TOOLS = [{"type": "function", "function": {"name": "read_file", "description": "Read a file.",
                                           "parameters": {"type": "object", "properties": {"filepath": {"type": "string"}}, "required": ["filepath"]}}}]
MSGS = [{"role": "system", "content": "You are a software engineer. Use the tools."}, {"role": "user", "content": "Read the file README.md."}]
g2 = {}
for name in ADAPTERS:
    errors, calls = 0, 0
    for i in range(10):
        try:
            calls += bool(chat(name, MSGS, TOOLS)["choices"][0]["message"].get("tool_calls"))
        except Exception as e:
            errors += 1
            print("request", i, "to", name, "failed:", str(e)[:300])
    g2[name] = errors == 0
    print(f"G2 {name}: 10 requests, {errors} errors, {calls} with a tool call -> " + ("PASS" if errors == 0 else "FAIL"))
probe = {}
for m in [SERVED_NAME] + list(ADAPTERS):
    lp = chat(m, MSGS, TOOLS, max_tokens=1, temperature=0, logprobs=True, top_logprobs=3)["choices"][0]["logprobs"]["content"][0]
    probe[m] = [(t["token"], round(t["logprob"], 6)) for t in lp["top_logprobs"]]
    print(m, "first token:", probe[m])
for name in ADAPTERS:
    same = probe[name] == probe[SERVED_NAME]
    g2[name] = g2[name] and not same
    print(f"G2 {name}: first-token log-probabilities " + ("IDENTICAL to the base model's: the adapter does not seem to be applied -> FAIL" if same else "differ from the base model's: the adapter is applied"))
```

## 7. The primary outcome: edit success on 300 held-out edit steps

Each step is a prefix of a held-out commit of the tune repositories (101 of the 137 files also occur in the training pairs; the commits do not), rendered the way the scorer renders the shipped agent, with thinking on. Every model writes
`EVAL_SAMPLES` generations; each call is applied to the window the model was shown with the harness's matcher. `good` = applied at the exact tier, changed, within 3 lines of the gold edit. The first model is the reference.

```python
PY = f"{VENV}/bin/python"
EP = f"{CODE_DIR}/train/edit_pairs.py"
os.makedirs(OUT_DIR, exist_ok=True)
eval_dir = f"{OUT_DIR}/edit_eval"
shutil.rmtree(eval_dir, ignore_errors=True)
rc, out = run_streaming([PY, "-u", EP, "eval", "--tokenizer", LOCAL_MODEL, "--base-url", base, "--data", LOCAL_STATES, "--seed", str(SEED), "--workers", str(WORKERS),
                         "--n", str(EVAL_N), "--samples", str(EVAL_SAMPLES), "--out", eval_dir, "--model", SERVED_NAME, *sum([["--model", a] for a in ARMS], [])])
assert rc == 0, "edit evaluation failed (output above)"
```

## 8. G3, G4, G4b, G5 and the format probe

`train/s4_gates.py` sends every check prefix to the base model and to each adapter as token ids (shipped sampling, up to 8,192 tokens) and prints one verdict per gate: PASS, FAIL, INSUFFICIENT (too little data) or INVALID (requests failed).
**Only PASS passes.** Exit code 0 = every gate passed, 1 = a gate failed or there was too little data, 2 = INVALID. The probe then re-samples the prefixes before a call the framework refused (report only, no threshold).
This is the slow cell; it prints a line after each model.

```python
gates_dir = f"{OUT_DIR}/gates"
shutil.rmtree(gates_dir, ignore_errors=True)
gate_rc, out = run_streaming([sys.executable, f"{CODE_DIR}/train/s4_gates.py", "--check", f"{CHECK_DIR}/check.jsonl", "--failed-edit", f"{CHECK_DIR}/check_failed_edit.jsonl",
                              "--base-url", base, "--model", SERVED_NAME, *sum([["--model", n] for n in ADAPTERS], []), "--out", gates_dir])
print("\ns4_gates exit code", gate_rc, "(0 = every gate passed, 1 = a gate failed or too little data, 2 = INVALID: requests failed)")
if gate_rc == 2:
    print("*** INVALID. vLLM server exit code (None = still running):", server.poll())
    print("".join(open("/content/vllm.log").readlines()[-60:]))
shutil.copy(f"{CHECK_DIR}/stats.json", f"{gates_dir}/check_stats.json")
probe_rc, out = run_streaming([sys.executable, f"{CODE_DIR}/train/format_probe.py", "--rejected", f"{CHECK_DIR}/check_rejected.jsonl", "--base-url", base,
                               "--model", SERVED_NAME, *sum([["--model", n] for n in ADAPTERS], []), "--out", f"{OUT_DIR}/probe"])
print("format probe exit code", probe_rc)
print("G2:", g2)
```

## 9. The pre-registered reading, the files to keep, and the end

`decide` applies the rules written down in `docs/PLAN_EDIT_PAIRS.md` before any sample existed: the good-edit rate at least 10 points above the base with the 95% interval above 0, `reproduces` not more than 2 points lower,
the share of steps with an edit call within 10 points of the base's, no more unreadable calls than the base, and all four gates PASS. Whatever it prints is what the paper says; a negative result is reported as one.

```python
rc, out = run_streaming([PY, EP, "decide", "--eval", f"{eval_dir}/eval_summary.json", "--gates", f"{gates_dir}/summary.json", "--ref", SERVED_NAME, "--out", f"{OUT_DIR}/decision.json"])
dst = DRIVE_OUT
shutil.copytree(OUT_DIR, dst, dirs_exist_ok=True)
print("\ncopied to", dst)
```
